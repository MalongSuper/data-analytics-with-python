```python
import math

def heron_formula(a, b, c):
    p = (a + b + c) / 2  # semiperimeter
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    return s

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))
print("Area of the triangle:", heron_formula(a, b, c))
```

```python
# Task: Determine if a number is a prime number.
def is_prime(n):
    if n <= 1:
        print(n, "is not a prime number")
        return
    elif n == 2:
        print(n, "is a prime number")
        return
    else:
        for i in range(2, (n // 2) + 1):
            if n % i == 0:
                print(n, "is not a prime number")
                return
        print(n, "is a prime number")
        return

number = int(input("Enter a number: "))
is_prime(number)
```

```python
# Task: Check whether a given number is positive, negative, or zero.
def check_sign(n):
    if n > 0:
        print(n, "is positive")
    elif n < 0:
        print(n, "is negative")
    else:
        print("The number is", n)

n = float(input("Enter a number: "))
check_sign(n)
```

```python
# Task: Check if a substring exists within a main string.
def check_substring(s1, s2):
    if s2 in s1:
        print("String s2 is in string s1")
    else:
        print("String s2 is not in string s1")

s1 = str(input("Enter string 1: "))
s2 = str(input("Enter string 2: "))
check_substring(s1, s2)
```

```python
# Task: Categorize a performance grade based on a numeric score.
def check_score(n):
    if 0 <= n < 5:
        print("Fail")
    elif 5 <= n < 7:
        print("Pass")
    elif 7 <= n < 9:
        print("Good")
    elif 9 <= n < 10:
        print("Excellent")
    elif n == 10:
        print("Outstanding")
    else:
        print("Invalid score")

n = float(input("Enter score: "))
check_score(n)
```

```python
# Task: Determine if a given year is a leap year.
def is_leap_year(y):
    if (y % 4 == 0) and (y % 100 != 0 or y % 400 == 0):
        print(y, "is a leap year")
    else:
        print(y, "is not a leap year")

year = int(input("Enter a year: "))
if year < 0:
    print("Invalid year")
else:
    is_leap_year(year)
```

```python
# Task: Count occurrences of a specific character in a string.
def count_letter(s, c):
    count = 0
    for i in range(len(s)):
        if s[i] == c:
            count += 1
    return count

string = str(input("Enter a string: "))
c = str(input("Enter a character: "))
print("The character appears", count_letter(string, c), "times")
```

```python
# Task: Generate the Fibonacci sequence up to n elements.
def fibonacci(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter the number of elements: "))
fib_sequence = [fibonacci(i) for i in range(1, n + 1)]
print("Fibonacci sequence with", n, "elements:", fib_sequence)
```

```python
# Task: Check if a string is a palindrome.
def palindrome(s):
    if s == s[::-1]:
        print("The string is a palindrome")
    else:
        print("The string is not a palindrome")

string = str(input("Enter a string: "))
palindrome(string)
```

### 1. Introduction to Data Analysis with Pandas & Matplotlib

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sample_data = {'Name': ['Admed', 'Omar', 'Ali', 'Ziad', 'Salwa', 'Lila'],
               'Mobile_Sales': [2540, 1370, 1320, 2000, 2100, 2150],
               'TV_Sales': [2200, 1900, 2150, 1850, 1770, 2000]}

df = pd.DataFrame(sample_data)
print(df)
```

```python
# Bar plot
categories = df['Name'].tolist()
mobile_sales_group = df['Mobile_Sales'].tolist()
tv_sales_group = df['TV_Sales'].tolist()

# Bar position in x-axis
x = np.arange(len(categories))
bar_width = 0.35  # The width of the bar

# Draw bar plot
# If we use:
# plt.bar(x, mobile_sales_group, width=bar_width, label='Mobile_Sales')
# plt.bar(x, tv_sales_group, width=bar_width, label='TV_Sales')
# The output will be two overlapping bars
plt.bar(x - bar_width/2, mobile_sales_group, width=bar_width, label='Mobile_Sales')
plt.bar(x + bar_width/2, tv_sales_group, width=bar_width, label='TV_Sales')

# Add labels and title
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Sales Volume of two salesman in January and April 2025')
plt.xticks(x, categories)
plt.legend()

plt.show()
```

```python
# Pie Chart
labels = df['Name'].tolist()
colors = ['blue', 'orange', 'green', 'red', 'purple', 'brown']

mobile_sales_group = df['Mobile_Sales'].tolist()
# Calculate the percentage
mobile_sales_sizes = [(mobile_sales_group[i] / sum(mobile_sales_group)) * 100
                      for i in range(len(mobile_sales_group))]
print("Mobile Sales (%):", mobile_sales_sizes)

tv_sales_group = df['TV_Sales'].tolist()
tv_sales_sizes = [(tv_sales_group[i] / sum(tv_sales_group)) * 100
                      for i in range(len(tv_sales_group))]
print("TV Sales (%):", tv_sales_sizes)

# Draw two pie charts using subplots
fig, axs = plt.subplots(1, 2, figsize=(8, 4))
axs[0].pie(mobile_sales_sizes,  labels=labels, colors=colors,
        shadow=False, startangle=90)
axs[1].pie(tv_sales_sizes,  labels=labels, colors=colors,
        shadow=False, startangle=90)
# Add a title
axs[0].set_title('Mobile_Sales', x=-0.2, y=0.2, rotation=90)
axs[0].legend()
axs[1].set_title('TV_Sales', x=-0.2, y=0.2, rotation=90)
axs[1].legend()
# Increase the gap between the two pie charts
plt.subplots_adjust(wspace=0.4)
plt.suptitle('Sales Volume of two salesman in January and April 2025')
plt.show()
```

```python
import string

text = ("Egypt is a developing country with the second-largest economy in Africa. "
        "It is considered to be a regional power in the Middle East, North Africa and the Muslim world, "
        "and a middle power worldwide. "
        "Islam is the official religion and Arabic is official language.")

text = text.translate(str.maketrans('', '', string.punctuation))

# Write the file
with open("Egypt.txt", "w") as f:
    f.write(text.lower())

text_list = []  # The list to obtain the word from the text
text_dict = {}  # The dictionary
with open("Egypt.txt", "r") as read_file:
    for read in read_file:
        # Iterate through each text in the file
        for text in read.split():
            text_list.append(text)

# Iterate through each word in the list to find its occurrence
for i in text_list:
    # Append to the dictionary
    text_dict[i] = text_list.count(i)

# Show the dictionary
print("All words and their occurrence")
for key, values in text_dict.items():
    print(f"+ Word: {key}, Count: {values}")

# Get max values
max_values = max(text_dict.values())
# Find the word with most count and return it
print("Words with maximum occurrence")
for key, values in text_dict.items():
    if values == max_values:
        print(f"+ Word: {key}, Count: {values}")
```

```python
stats = ['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max']
data = {}

data['count'] = [df['Age'][df['Animals'] == i].count()
                 for i in df['Animals'].unique()]
data['mean'] = [df['Age'][df['Animals'] == i].mean()
                for i in df['Animals'].unique()]
data['std'] = [df['Age'][df['Animals'] == i].std()
               for i in df['Animals'].unique()]
data['min'] = [df['Age'][df['Animals'] == i].min()
               for i in df['Animals'].unique()]
data['25%'] = [df['Age'][df['Animals'] == i].quantile(0.25)
               for i in df['Animals'].unique()]
data['50%'] = [df['Age'][df['Animals'] == i].quantile(0.5)
               for i in df['Animals'].unique()]
data['75%'] = [df['Age'][df['Animals'] == i].quantile(0.75)
               for i in df['Animals'].unique()]
data['max'] = [df['Age'][df['Animals'] == i].max()
               for i in df['Animals'].unique()]

df = pd.DataFrame(data, index=df['Animals'].unique())
df
```

## 4. Advanced Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

cities = ['A', 'B', 'C', 'D', 'E', 'F']
data = {}
n_samples = 500
# first value is 0, rest are cumulative sum
# Cumulative sum is used as the sum of all the previous elements
# including the current one
# E.g., a1, a1 + a2, a1 + a2, a3, ...
# This means the new temperature depends on the previous temperatures
for i in cities:
    # With this, the next temperature might slightly increase or decrease
    # based on the previous temperature
    deltas = np.random.uniform(-1, 1, n_samples)
    # Cumulative sum
    temps = np.cumsum(deltas) # Return an array
    data[i] = temps

# Put into the dataframe
df = pd.DataFrame(data)

df.plot()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Tips dataset
tips = sns.load_dataset("tips")

# Calculate tips percentage
tips['percentage_tip'] = round((tips['tip'] / tips['total_bill']) * 100)

# Filter data by gender and time (Lunch/Dinner)
male_lunch_tips = tips[(tips['sex'] == 'Male') & (tips['time'] == 'Lunch')]
female_lunch_tips = tips[(tips['sex'] == 'Female') & (tips['time'] == 'Lunch')]
male_dinner_tips = tips[(tips['sex'] == 'Male') & (tips['time'] == 'Dinner')]
female_dinner_tips = tips[(tips['sex'] == 'Female') & (tips['time'] == 'Dinner')]

# Create a figure with 2x2 subplots (4 histograms)
fig, axes = plt.subplots(2, 2, figsize=(10, 6))
# Set common Y-axis ticks
yticks = np.arange(0, 36, 5)
xticks = np.arange(0, 40, 5)

# Plot for male lunch tips
sns.histplot(male_lunch_tips['percentage_tip'], color='blue', ax=axes[0, 0],
             label='Male Lunch', alpha=0.6)
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_yticks(yticks)
axes[0, 0].set_xticks(xticks)
axes[0, 0].legend()

# Plot for female lunch tips
sns.histplot(female_lunch_tips['percentage_tip'], color='blue', ax=axes[1, 0],
             label='Female Lunch', alpha=0.6)
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_yticks(yticks)
axes[1, 0].set_xticks(xticks)
axes[1, 0].legend()

# Plot for male dinner tips
sns.histplot(male_dinner_tips['percentage_tip'], color='blue', ax=axes[0, 1],
             label='Male Dinner', alpha=0.6)
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].set_yticks(yticks)
axes[0, 1].set_xticks(xticks)
axes[0, 1].legend()

# Plot for female dinner tips
sns.histplot(female_dinner_tips['percentage_tip'], color='blue', ax=axes[1, 1],
             label='Female Dinner', alpha=0.6)
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_yticks(yticks)
axes[1, 1].set_xticks(xticks)
axes[1, 1].legend()

# Set common x-labels
for ax in axes.flat:
    ax.set_xlabel('Percentage of Tip')

# Adjust layout for better visualization
plt.tight_layout()

# Show plot
plt.show()
```
